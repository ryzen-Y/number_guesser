## Part 1: Reading Text Files

**Key Concepts:**

| Keyword / Method | Description |
|------------------|-------------|
| `open()` | Built-in function to open a file |
| `'r'` | Read mode (default) |
| `'w'` | Write mode (overwrites existing content) |
| `'a'` | Append mode (adds to end of file) |
| `with` | Context manager — auto-closes file when done |
| `.read()` | Read entire file content as a single string |
| `.readlines()` | Read all lines into a list |

````python
# First, create a sample file to work with
with open('sample.txt', 'w') as f:
    f.write('Hello, World!\n')
    f.write('This is line 2.\n')
    f.write('And this is line 3.\n')

print('sample.txt created.')
````

````python
# Read entire file as a single string
with open('sample.txt', 'r') as f:
    content = f.read()
    print(content)
````

````python
# Read file line by line
with open('sample.txt', 'r') as f:
    for line in f:
        print(line.strip())   # .strip() removes trailing newline
````

````python
# Read all lines into a list
with open('sample.txt', 'r') as f:
    lines = f.readlines()

print(lines)              # list with '\n' at end of each item
print(f'Total lines: {len(lines)}')
````

## Part 2: Writing Text Files

> **Best Practice:** Always use the `with` statement. It ensures the file is properly closed even if an error occurs, preventing data loss and resource leaks.

````python
# Write mode — creates or overwrites the file
with open('output.txt', 'w') as f:
    f.write('Hello, World!\n')
    f.write('Second line\n')

print('output.txt written.')
````

````python
# Write a list of lines with writelines()
lines = ['Alice\n', 'Bob\n', 'Carol\n']
with open('names.txt', 'w') as f:
    f.writelines(lines)

print('names.txt written.')
````

````python
# Append mode — adds to end without overwriting
with open('log.txt', 'a') as f:
    f.write('New entry added\n')

# Loop and write each item
data = ['row1', 'row2', 'row3']
with open('data_out.txt', 'w') as f:
    for item in data:
        f.write(item + '\n')

print('Files written.')
````

## Part 3: Working with CSV Files

**CSV (Comma-Separated Values):**
- Plain text, tabular data
- Each row is a new line; values separated by commas
- First row is usually the header
- Python's built-in `csv` module handles reading and writing cleanly

````python
import csv

# Create a sample CSV first
with open('students.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['name', 'age', 'score'])
    writer.writerow(['Alice', 22, 88])
    writer.writerow(['Bob', 21, 74])
    writer.writerow(['Carol', 23, 91])
    writer.writerow(['Dave', 20, 60])

print('students.csv created.')
````

````python
# Reading CSV with csv.reader
with open('students.csv', 'r') as f:
    reader = csv.reader(f)
    header = next(reader)   # skip header row
    print('Header:', header)
    for row in reader:
        print(row)
````

````python
# Reading CSV as dictionary rows (more readable)
with open('students.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"{row['name']} — age: {row['age']}, score: {row['score']}")
````

````python
# Writing CSV with csv.DictWriter
summary = [
    {'file': 'a.csv', 'rows': 100},
    {'file': 'b.csv', 'rows': 250},
]
with open('summary.csv', 'w', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=['file', 'rows'])
    writer.writeheader()
    writer.writerows(summary)

print('summary.csv written.')
````

## Part 4: Working with JSON Files

**When to use JSON vs CSV:**
- **CSV** — Tabular data, spreadsheets, simple row/column data
- **JSON** — Nested/hierarchical data, APIs, config files, mixed types

````python
import json

# Writing JSON to a file
student = {'name': 'Alice', 'age': 22, 'grades': [88, 91, 79]}
with open('student.json', 'w') as f:
    json.dump(student, f, indent=4)

print('student.json written.')
````

````python
# Reading JSON from a file
with open('student.json', 'r') as f:
    data = json.load(f)

print(data)
print('Name:', data['name'])
print('Grades:', data['grades'])
````

````python
# Parsing a JSON string (common when working with APIs)
text = '{"name": "Alice", "age": 22}'
obj = json.loads(text)
print(obj['name'])   # Alice

# Converting a Python object to JSON string
output = json.dumps(student, indent=2)
print(output)
````

## Part 6: Using the `glob` Module

**glob** finds files matching a wildcard pattern:

| Pattern | Matches |
|---------|--------|
| `*.txt` | All `.txt` files in current folder |
| `data/*.csv` | All CSVs inside the `data/` folder |
| `**/*.json` | JSON files in all subdirectories (recursive) |
| `file_?.csv` | `file_1.csv`, `file_2.csv`, etc. |
| `[0-9]*.txt` | Files starting with a digit |

````python
import glob

# Find all .txt files in the current directory
txt_files = glob.glob('*.json')
print('TXT files:', txt_files)
````

## 🎯 Assignment 1: Patient Data Pipeline

**Task:** Build an end-to-end file handling pipeline that processes patient CSV files.

**Steps:**
1. Create a structured project folder: `data/raw`, `data/processed`, `output`
2. Create 3 sample patient CSV files and place them in `data/raw`
3. Write a script that reads each CSV, counts patients per file, and filters rows where `age > 50`
4. Save filtered data per file into `data/processed` as new CSVs
5. Generate a summary JSON in `output/` with filename, total rows, and filtered count
6. ⭐ **Bonus:** Use `glob` to auto-detect all CSVs without hardcoding filenames

````python
# ── ASSIGNMENT 1 SOLUTION ─────────────────────────────────────────────────────
import os, glob, csv, json

# Step 1: Create folder structure
for folder in ['assignment/data/raw', 'assignment/data/processed', 'assignment/output']:
    os.makedirs(folder, exist_ok=True)
print('Step 1 ✓ — Folder structure created.')
````

````python
# Step 2: Create 3 sample patient CSV files
import random
random.seed(42)

names = ['Alice','Bob','Carol','Dave','Eve','Frank','Grace','Hank','Iris','Jack',
         'Karen','Leo','Mia','Nate','Olivia','Paul','Quinn','Rose','Sam','Tina']

for i in range(1, 4):
    filepath = f'assignment/data/raw/patients_{i}.csv'
    with open(filepath, 'w', newline='') as f:
        writer = csv.DictWriter(f, fieldnames=['id', 'name', 'age', 'diagnosis'])
        writer.writeheader()
        for j in range(1, 9):
            writer.writerow({
                'id': j,
                'name': random.choice(names),
                'age': random.randint(20, 80),
                'diagnosis': random.choice(['Diabetes', 'Hypertension', 'Healthy', 'Asthma'])
            })
    print(f'  Created: {filepath}')

print('Step 2 ✓ — Sample patient files created.')
````

````python
# Steps 3, 4 & 5 (+ Bonus): Process all CSVs using glob, filter age > 50, save outputs
summary = []

# ⭐ Bonus: auto-detect all CSVs using glob (no hardcoded filenames)
for fp in glob.glob('assignment/data/raw/*.csv'):
    fname = os.path.basename(fp)

    # Step 3: Read and filter
    with open(fp, 'r') as f:
        all_patients = list(csv.DictReader(f))

    filtered = [p for p in all_patients if int(p['age']) > 50]

    # Step 4: Save filtered data to data/processed/
    out_path = os.path.join('assignment/data/processed', fname)
    with open(out_path, 'w', newline='') as f:
        writer = csv.DictWriter(f, fieldnames=['id', 'name', 'age', 'diagnosis'])
        writer.writeheader()
        writer.writerows(filtered)

    # Step 5: Collect summary info
    summary.append({
        'file': fname,
        'total_rows': len(all_patients),
        'filtered_count': len(filtered)
    })
    print(f'  {fname}: {len(all_patients)} total, {len(filtered)} aged > 50')

print('Steps 3 & 4 ✓ — Files processed and saved to data/processed/')
````

````python
# Step 5: Save summary JSON
with open('assignment/output/summary.json', 'w') as f:
    json.dump(summary, f, indent=2)

print('Step 5 ✓ — Summary saved to assignment/output/summary.json')
print()

# Final verification
with open('assignment/output/summary.json', 'r') as f:
    print(json.dumps(json.load(f), indent=2))
````
